```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
# In this exercise I will only create a function which includes two built-in functions


def canny_edge_detection(image, sigma, low_threshold, high_threshold):
    # Gaussian smoothing
    # Kernel size is automatically determined from sigma
    kernel_size = int(2 * np.ceil(3 * sigma) + 1)
    blurred = cv2.GaussianBlur(image, (kernel_size, kernel_size), sigma)
    edges = cv2.Canny(blurred, low_threshold, high_threshold)
    return edges

# testing with provided image:

if __name__ == "__main__":
    img_color = cv2.imread('.\Canny_img.png')
    
    # Converting to grayscale
    # Note that this operation is very important for detecting image edges
    img_gray = cv2.cvtColor(img_color, cv2.COLOR_BGR2GRAY)

    # Parameters
    sigma = 1.1
    low_threshold = 50
    high_threshold = 150
    
    # Detect edges
    edges = canny_edge_detection(img_gray, sigma, low_threshold, high_threshold)

    
    # cv2.imwrite('input_grayscale.jpg', img_gray)
    # cv2.imwrite('output_edges.jpg', edges)
    
    # Displaying results using subplot
    plt.figure(figsize=(12, 6))
    plt.subplot(1, 2, 1)
    plt.imshow(img_gray, cmap='gray')
    plt.title('Grayscale Input')
    plt.axis('off')
    
    plt.subplot(1, 2, 2)
    plt.imshow(edges, cmap='gray')
    plt.title('Canny Edges')
    plt.axis('off')
    
    plt.show()
```
